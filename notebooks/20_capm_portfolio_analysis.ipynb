# CAPM & Portfolio Risk Analysis

**Portfolio category:** Technical demonstration

Estimate beta, alpha and Sharpe ratio from simulated market and asset returns and compare a small portfolio.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
rng=np.random.default_rng(20); n=756
dates=pd.bdate_range('2023-01-02',periods=n); market=rng.normal(.0003,.009,n)
assets=pd.DataFrame(index=dates)
for name,beta,sig in [('Asset_A',1.1,.006),('Asset_B',.7,.008),('Asset_C',1.4,.011)]:
 assets[name]=.0001+beta*market+rng.normal(0,sig,n)
rf=.00008
rows=[]
for c in assets:
 x=market; y=assets[c]; beta=np.cov(x,y,ddof=1)[0,1]/np.var(x,ddof=1); alpha=(y-rf).mean-beta*(x-rf).mean(); sharpe=(y-rf).mean()/y.std()*np.sqrt(252); rows.append([c,beta,alpha,sharpe])
metrics=pd.DataFrame(rows,columns=['Asset','Beta','Daily_Alpha','Annualized_Sharpe']); display(metrics.round(4))
w=np.array([.4,.35,.25]); port=assets@w; print('Portfolio annualized return:',round(port.mean()*252,3)); print('Portfolio annualized volatility:',round(port.std()*np.sqrt(252),3)); print('Portfolio Sharpe:',round((port.mean()-rf)*np.sqrt(252)/port.std(),3))
fig,ax=plt.subplots(figsize=(9,4)); (1+port).cumprod().plot(ax=ax); ax.set_title('Simulated Portfolio Growth'); plt.tight_layout()
```
