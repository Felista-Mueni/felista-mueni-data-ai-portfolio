# RetailMart Kenya — EDA & Dashboard Preparation

Synthetic portfolio data is generated below. The workflow demonstrates KPI analysis, segmentation and dashboard preparation.

```python
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
n=1200
df=pd.DataFrame({'Region':rng.choice(['Nairobi','Central','Coast','Rift Valley','Western','Eastern'],n),'Category':rng.choice(['Electronics','Grocery','Home','Beauty','Fashion'],n),'Quantity':rng.integers(1,8,n),'Unit_Price':rng.choice([450,750,1200,2500,5000,8500,15000,25000,45000],n)})
df['Revenue']=df.Quantity*df.Unit_Price
df['Discount']=rng.uniform(0,.18,n)
df['Net_Revenue']=df.Revenue*(1-df.Discount)
df['Profit']=df.Net_Revenue*rng.uniform(.18,.45,n)
df.head()
```

```python
kpis={'Orders':len(df),'Net Revenue':df.Net_Revenue.sum(),'Profit':df.Profit.sum(),'Margin %':100*df.Profit.sum()/df.Net_Revenue.sum()}
pd.Series(kpis)
```

```python
region=df.groupby('Region')['Net_Revenue'].sum().sort_values(ascending=False)
region.plot(kind='bar',title='Net Revenue by Region'); plt.tight_layout()
```

## Dashboard recommendations
KPI cards, monthly trend, regional/category performance, profitability and filters for date, region, category and customer segment.
