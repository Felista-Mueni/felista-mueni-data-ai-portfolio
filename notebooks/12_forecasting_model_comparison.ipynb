# Forecasting Model Comparison: ARIMA vs Random Forest

**Portfolio category:** Professional technical demonstration

Time-aware forecasting demonstration comparing ARIMA with a lag-feature Random Forest on synthetic monthly demand data.

> This notebook uses simulated data and is clearly labelled as portfolio evidence.


```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error
rng=np.random.default_rng(7); dates=pd.date_range('2019-01-01',periods=84,freq='MS'); t=np.arange(84)
ts=pd.Series(100+.7*t+18*np.sin(2*np.pi*t/12)+rng.normal(0,7,84),index=dates,name='demand')
train,test=ts.iloc[:-12],ts.iloc[-12:]
ar=ARIMA(train,order=(2,1,2)).fit(); ar_pred=ar.forecast(12)
def lags(s,L=12):
 d=pd.DataFrame({'y':s})
 for i in range(1,L+1): d[f'lag_{i}']=d.y.shift(i)
 d['month']=d.index.month
 return d.dropna()
d=lags(ts); tr=d[d.index<test.index[0]]; te=d[d.index>=test.index[0]]; features=[c for c in d if c!='y']
rf=RandomForestRegressor(n_estimators=400,max_depth=8,min_samples_leaf=2,random_state=42).fit(tr[features],tr.y); rf_pred=rf.predict(te[features])
comparison=pd.DataFrame({'Model':['ARIMA','Random Forest'],'MAE':[mean_absolute_error(test,ar_pred),mean_absolute_error(te.y,rf_pred)],'RMSE':[mean_squared_error(test,ar_pred)**.5,mean_squared_error(te.y,rf_pred)**.5]}); display(comparison.round(2))
fig,ax=plt.subplots(figsize=(9,5)); ax.plot(train,label='Train'); ax.plot(test,label='Actual'); ax.plot(test.index,ar_pred,label='ARIMA'); ax.plot(te.index,rf_pred,label='Random Forest'); ax.set_title('Out-of-Sample Forecast Comparison'); ax.legend(); plt.tight_layout()
```
