# AI Model Evaluation Framework

A model-agnostic rubric for evaluating AI-generated responses.

```python
import pandas as pd

rubric = pd.DataFrame([
    ["Relevance", "Directly addresses the instruction", 20],
    ["Factuality", "Accurate or appropriately qualified claims", 25],
    ["Instruction following", "Meets format and constraints", 20],
    ["Clarity", "Clear and coherent", 15],
    ["Safety/quality", "Avoids problematic content", 10],
    ["Consistency", "Stable terminology and reasoning", 10],
], columns=["Criterion", "Definition", "Weight"])

rubric
```

## Evaluation workflow

1. Read the task specification.
2. Identify required behaviour.
3. Score each criterion.
4. Record evidence for low scores.
5. Flag edge cases.
6. Re-check before submission.
