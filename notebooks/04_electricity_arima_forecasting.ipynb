# Electricity Generation — ARIMA Forecasting

Synthetic monthly series demonstrating a practical time-series workflow.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error

rng = np.random.default_rng(42)
dates = pd.date_range("2015-01-01", "2024-12-01", freq="MS")
t = np.arange(len(dates))
s = pd.Series(
    12000 + 55 * t + 1100 * np.sin(2 * np.pi * t / 12) + rng.normal(0, 450, len(t)),
    index=dates
)

train = s.iloc[:-12]
test = s.iloc[-12:]

fit = ARIMA(train, order=(1, 1, 1)).fit()
forecast = fit.forecast(12)

print("MAE:", mean_absolute_error(test, forecast))

plt.figure(figsize=(10, 5))
plt.plot(train, label="Train")
plt.plot(test, label="Actual")
plt.plot(forecast, label="Forecast")
plt.legend()
plt.title("Electricity Generation Forecast")
plt.show()
```

## Modelling caution

Compare multiple models and use rolling validation where appropriate. History length, seasonality and structural change materially affect forecasts.
