# Johansen Cointegration Analysis

**Portfolio category:** Technical demonstration

Demonstrate multivariate time-series analysis using Johansen trace statistics on simulated non-stationary series.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
from statsmodels.tsa.vector_ar.vecm import coint_johansen
rng=np.random.default_rng(17); n=300
x=np.cumsum(rng.normal(0,.8,n)); y=1.4*x+np.cumsum(rng.normal(0,.25,n)); z=-.7*x+np.cumsum(rng.normal(0,.35,n))
df=pd.DataFrame({'Series_A':x,'Series_B':y,'Series_C':z})
fig,ax=plt.subplots(figsize=(9,4)); df.plot(ax=ax); ax.set_title('Simulated Non-Stationary Series'); plt.tight_layout()
res=coint_johansen(df,det_order=0,k_ar_diff=1)
table=pd.DataFrame({'Trace Statistic':res.lr1,'5% Critical Value':res.cvt[:,1]},index=['r=0','r<=1','r<=2'])
table['Reject_5pct']=table['Trace Statistic']>table['5% Critical Value']; display(table.round(3))
print('The Johansen procedure tests for the number of cointegrating relationships in the system.')
```
