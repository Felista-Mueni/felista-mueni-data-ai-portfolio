# PRISMA Screening Workflow

Synthetic systematic-review screening flow with deduplication and transparent exclusion accounting.

**Evidence status:** Reproducible synthetic/illustrative portfolio demonstration; not client data or externally validated outcomes.


```python
import pandas as pd,numpy as np,matplotlib.pyplot as plt
rng=np.random.default_rng(25)
records=pd.DataFrame({'record_id':range(1,401),'title':['Study '+str(i) for i in range(1,401)],'duplicate':rng.random(400)<.12})
unique=records.loc[~records.duplicate].copy(); unique['title_abstract_include']=rng.random(len(unique))<.32
full=unique.loc[unique.title_abstract_include].copy(); full['full_text_available']=rng.random(len(full))<.9; full['eligible']=rng.random(len(full))<.48
identified=len(records); duplicates=int(records.duplicate.sum()); screened=len(unique); full_sought=len(full); unavailable=int((~full.full_text_available).sum()); assessed=int(full.full_text_available.sum()); included=int((full.full_text_available & full.eligible).sum())
flow=pd.DataFrame({'stage':['Identified','Duplicates removed','Title/abstract screened','Full texts sought','Full texts unavailable','Full texts assessed','Studies included'],'count':[identified,duplicates,screened,full_sought,unavailable,assessed,included]})
display(flow); print('Full-text exclusions:',assessed-included)
fig,ax=plt.subplots(figsize=(9,4)); ax.barh(flow.stage,flow['count']); ax.invert_yaxis(); ax.set_title('Illustrative Screening Counts — Synthetic'); plt.tight_layout()
print('This is a teaching demonstration, not an actual completed PRISMA report.')
```
