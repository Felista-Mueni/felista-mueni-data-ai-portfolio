# House Prices — XGBoost Predictive Modelling

Synthetic data demonstrates a complete supervised-learning workflow.

```python
import numpy as np, pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
from xgboost import XGBRegressor
rng=np.random.default_rng(42); n=1200
df=pd.DataFrame({'OverallQual':rng.integers(3,11,n),'GrLivArea':rng.normal(1500,450,n).clip(500,4000),'GarageCars':rng.integers(0,4,n),'TotalBsmtSF':rng.normal(1000,350,n).clip(0,2500),'BedroomAbvGr':rng.integers(1,6,n),'YearBuilt':rng.integers(1950,2022,n)})
df['SalePrice']=(25000+df.OverallQual*18000+df.GrLivArea*75+df.GarageCars*12000+df.TotalBsmtSF*35+df.YearBuilt*280+rng.normal(0,30000,n)).clip(30000,None)
X=df.drop(columns='SalePrice'); y=df.SalePrice
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
model=XGBRegressor(n_estimators=400,max_depth=5,learning_rate=.04,subsample=.85,colsample_bytree=.85,random_state=42)
model.fit(Xtr,ytr); pred=model.predict(Xte)
print('RMSE:',mean_squared_error(yte,pred)**.5,'R2:',r2_score(yte,pred))
```
