# Quantitative Finance — Principal Component Analysis

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

rng = np.random.default_rng(42)
z = rng.normal(size=(800, 3))

X = pd.DataFrame({
    "Factor_A": z[:, 0] + 0.2 * z[:, 1],
    "Factor_B": 0.8 * z[:, 0] + 0.4 * z[:, 1],
    "Factor_C": z[:, 1] + 0.2 * z[:, 2],
    "Factor_D": 0.5 * z[:, 0] + 0.7 * z[:, 2],
    "Factor_E": z[:, 2] + 0.1 * z[:, 0]
})

Z = StandardScaler().fit_transform(X)
pca = PCA().fit(Z)

results = pd.DataFrame({
    "PC": range(1, 6),
    "Explained_Variance": pca.explained_variance_ratio_,
    "Cumulative": np.cumsum(pca.explained_variance_ratio_)
})

results
```

PCA transforms correlated variables into orthogonal components ranked by explained variance.
