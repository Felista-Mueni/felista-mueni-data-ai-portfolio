# Heart Disease Classification

**Portfolio category:** Machine Learning / Classification

Synthetic portfolio demonstration of preprocessing, model comparison, ROC-AUC and error analysis. This is not a clinical model.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,ConfusionMatrixDisplay,RocCurveDisplay
rng=np.random.default_rng(42); n=1200
age=rng.normal(54,10,n).clip(25,85); bp=rng.normal(132,18,n).clip(85,210); chol=rng.normal(235,45,n).clip(120,420); hr=(190-age*.7+rng.normal(0,18,n)).clip(70,210); oldpeak=np.abs(rng.normal(1.1,.9,n)).clip(0,5); angina=rng.binomial(1,.30,n); sex=rng.binomial(1,.55,n)
logit=-5+.055*age+.012*(chol-200)+.018*(bp-120)-.018*(hr-150)+.65*oldpeak+1*angina+.25*sex
p=1/(1+np.exp(-logit)); y=rng.binomial(1,p)
df=pd.DataFrame({'age':age,'resting_bp':bp,'cholesterol':chol,'max_hr':hr,'oldpeak':oldpeak,'exercise_angina':angina,'sex':sex,'heart_disease':y})
X=df.drop(columns='heart_disease'); y=df.heart_disease
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
models={'Logistic Regression':Pipeline([('imputer',SimpleImputer(strategy='median')),('scale',StandardScaler()),('model',LogisticRegression(max_iter=1000))]),'Random Forest':Pipeline([('imputer',SimpleImputer(strategy='median')),('model',RandomForestClassifier(n_estimators=300,max_depth=7,min_samples_leaf=4,random_state=42))])}
rows=[]
for name,m in models.items():
 m.fit(Xtr,ytr); pred=m.predict(Xte); prob=m.predict_proba(Xte)[:,1]; rows.append([name,accuracy_score(yte,pred),precision_score(yte,pred),recall_score(yte,pred),f1_score(yte,pred),roc_auc_score(yte,prob)])
results=pd.DataFrame(rows,columns=['Model','Accuracy','Precision','Recall','F1','ROC_AUC']).sort_values('ROC_AUC',ascending=False)
display(results.round(3))
```

```python
fig,ax=plt.subplots(figsize=(7,5))
for name,m in models.items(): RocCurveDisplay.from_estimator(m,Xte,yte,ax=ax,name=name)
ax.set_title('ROC Curve Comparison'); plt.tight_layout()
best=models[results.iloc[0].Model]
fig,ax=plt.subplots(figsize=(5,4)); ConfusionMatrixDisplay.from_estimator(best,Xte,yte,ax=ax); ax.set_title('Best Model Confusion Matrix'); plt.tight_layout()
```

## Evidence produced
- Reproducible synthetic dataset generation
- Stratified holdout validation
- Two model families
- Accuracy, precision, recall, F1 and ROC-AUC
- ROC and confusion-matrix visual evidence
- Clear distinction between portfolio evidence and clinical deployment
