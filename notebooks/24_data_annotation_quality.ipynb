# Data Annotation Quality Control

Synthetic annotation agreement, disagreement review and adjudication workflow.

**Evidence status:** Reproducible synthetic/illustrative portfolio demonstration; not client data or externally validated outcomes.


```python
import pandas as pd,numpy as np,matplotlib.pyplot as plt
from sklearn.metrics import cohen_kappa_score,confusion_matrix,ConfusionMatrixDisplay
rng=np.random.default_rng(24); n=200; truth=rng.integers(0,3,n); a=truth.copy(); b=truth.copy()
for labels in [a,b]:
 ix=rng.choice(n,size=28,replace=False); labels[ix]=rng.integers(0,3,len(ix))
df=pd.DataFrame({'item_id':range(1,n+1),'annotator_a':a,'annotator_b':b})
df['agree']=df.annotator_a==df.annotator_b
print('Agreement:',round(df.agree.mean(),3),'Cohen kappa:',round(cohen_kappa_score(a,b),3))
display(df[~df.agree].head(15))
ConfusionMatrixDisplay(confusion_matrix(a,b),display_labels=['A','B','C']).plot(); plt.title('Inter-annotator Agreement'); plt.tight_layout()
print('Disagreements require adjudication; neither annotator is automatically the ground truth.')
```
