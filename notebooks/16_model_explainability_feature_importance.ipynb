# Model Explainability & Feature Importance

**Portfolio category:** Technical demonstration

Compare permutation importance with built-in tree importance to explain model drivers without overstating causality.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
d=load_diabetes(as_frame=True); X=d.data; y=d.target
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
m=RandomForestRegressor(n_estimators=400,min_samples_leaf=3,random_state=42).fit(Xtr,ytr); pred=m.predict(Xte)
print({'MAE':round(mean_absolute_error(yte,pred),2),'RMSE':round(mean_squared_error(yte,pred)**.5,2),'R2':round(r2_score(yte,pred),3)})
built=pd.Series(m.feature_importances_,index=X.columns).sort_values()
perm=permutation_importance(m,Xte,yte,n_repeats=15,random_state=42)
perm_s=pd.Series(perm.importances_mean,index=X.columns).sort_values()
fig,ax=plt.subplots(figsize=(7,5)); built.plot(kind='barh',ax=ax); ax.set_title('Random Forest Feature Importance'); plt.tight_layout()
fig,ax=plt.subplots(figsize=(7,5)); perm_s.plot(kind='barh',ax=ax); ax.set_title('Permutation Importance on Test Set'); plt.tight_layout()
print('Interpretation: importance indicates predictive contribution, not causal effect.')
```
