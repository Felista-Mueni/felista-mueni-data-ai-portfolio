# Kalman Filter State Estimation

**Portfolio category:** Technical demonstration

Estimate a latent time-varying signal from noisy observations using a one-dimensional Kalman filter.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
rng=np.random.default_rng(18); n=250; t=np.arange(n)
true=50+.08*t+3*np.sin(t/18); obs=true+rng.normal(0,3.5,n)
q=.08; r=12; x=obs[0]; P=1.; estimates=[]
for z in obs:
 P=P+q; K=P/(P+r); x=x+K*(z-x); P=(1-K)*P; estimates.append(x)
df=pd.DataFrame({'True_State':true,'Observation':obs,'Kalman_Estimate':estimates})
fig,ax=plt.subplots(figsize=(10,4)); ax.plot(df.True_State,label='Latent state'); ax.scatter(t,df.Observation,s=8,alpha=.35,label='Noisy observations'); ax.plot(df.Kalman_Estimate,label='Kalman estimate'); ax.set_title('Kalman Filtering'); ax.legend(); plt.tight_layout()
print('Observation RMSE:',round(np.sqrt(np.mean((obs-true)**2)),3)); print('Filtered RMSE:',round(np.sqrt(np.mean((np.array(estimates)-true)**2)),3))
```
