# Heston Model — Monte Carlo Option Pricing

**Portfolio category:** Professional technical demonstration

Quantitative finance demonstration implementing stochastic volatility, correlated Brownian shocks, Monte Carlo pricing, confidence intervals and correlation sensitivity.

> This notebook uses simulated data and is clearly labelled as portfolio evidence.


```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
def heston(S0=100,K=100,T=1,r=.05,v0=.04,kappa=2,theta=.04,sigma=.35,rho=-.7,steps=252,n_paths=20000,seed=42):
 rng=np.random.default_rng(seed); dt=T/steps; S=np.full(n_paths,S0,dtype=float); v=np.full(n_paths,v0,dtype=float)
 for _ in range(steps):
  z1=rng.standard_normal(n_paths); z2=rng.standard_normal(n_paths); w2=rho*z1+np.sqrt(1-rho*rho)*z2; vp=np.maximum(v,0)
  v=np.maximum(v+kappa*(theta-vp)*dt+sigma*np.sqrt(vp*dt)*w2,0)
  S*=np.exp((r-.5*vp)*dt+np.sqrt(vp*dt)*z1)
 payoff=np.maximum(S-K,0); price=np.exp(-r*T)*payoff.mean(); se=np.exp(-r*T)*payoff.std(ddof=1)/np.sqrt(n_paths)
 return price,se,S
price,se,terminal=heston(); print('Call price:',round(price,4),'95% CI:',(round(price-1.96*se,4),round(price+1.96*se,4)))
fig,ax=plt.subplots(figsize=(8,4)); ax.hist(terminal,bins=70); ax.axvline(100,ls='--'); ax.set_title('Heston Terminal Price Distribution'); plt.tight_layout()
rows=[]
for rho in [-.9,-.7,-.5,-.3,0,.3]:
 p,se,_=heston(rho=rho,steps=126,n_paths=10000,seed=123); rows.append([rho,p,se])
sens=pd.DataFrame(rows,columns=['rho','Call_Price','MC_SE']); display(sens.round(4))
fig,ax=plt.subplots(figsize=(7,4)); ax.plot(sens.rho,sens.Call_Price,marker='o'); ax.set_xlabel('Correlation'); ax.set_ylabel('Call price'); ax.set_title('Heston Correlation Sensitivity'); plt.tight_layout()
```
