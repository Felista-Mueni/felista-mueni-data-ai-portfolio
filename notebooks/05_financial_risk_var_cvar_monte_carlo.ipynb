# Quantitative Finance — VaR, CVaR & Monte Carlo

Synthetic returns demonstrate portfolio risk measures.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
r = pd.Series(rng.normal(0.007, 0.045, 120))

q = np.quantile(r, 0.05)
VaR95 = -q
CVaR95 = -r[r <= q].mean()
Sharpe = (r.mean() / r.std()) * np.sqrt(12)

print({
    "VaR_95": VaR95,
    "CVaR_95": CVaR95,
    "Annualized_Sharpe": Sharpe
})

sim = rng.normal(r.mean(), r.std(), 10000)

plt.figure(figsize=(10, 5))
plt.hist(sim, bins=50)
plt.axvline(-VaR95, linestyle="--", label="95% VaR")
plt.title("Simulated Returns")
plt.legend()
plt.show()
```

VaR gives a percentile loss threshold; CVaR summarizes the average loss beyond that threshold. Results depend on distributional assumptions and should be stress-tested.
