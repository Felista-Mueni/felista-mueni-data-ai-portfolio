# SQL Business Analytics

Synthetic retail SQL case: joins, CTEs, window functions and revenue analysis.

**Evidence status:** Reproducible synthetic/illustrative portfolio demonstration; not client data or externally validated outcomes.


```python
import sqlite3,pandas as pd,numpy as np
rng=np.random.default_rng(21); db=sqlite3.connect(':memory:')
customers=pd.DataFrame({'customer_id':range(1,101),'region':rng.choice(['Nairobi','Mombasa','Kisumu'],100)})
orders=pd.DataFrame({'order_id':range(1,1001),'customer_id':rng.integers(1,101,1000),'amount':rng.uniform(200,12000,1000).round(2),'order_date':pd.date_range('2025-01-01',periods=1000,freq='8h').astype(str)})
customers.to_sql('customers',db,index=False); orders.to_sql('orders',db,index=False)
query="""WITH regional AS (SELECT c.region, c.customer_id, SUM(o.amount) revenue, COUNT(*) orders FROM customers c JOIN orders o ON c.customer_id=o.customer_id GROUP BY c.region,c.customer_id) SELECT region,customer_id,ROUND(revenue,2) revenue,orders, DENSE_RANK() OVER(PARTITION BY region ORDER BY revenue DESC) region_rank FROM regional ORDER BY region,region_rank"""
display(pd.read_sql_query(query,db).head(20))
display(pd.read_sql_query("SELECT c.region,COUNT(*) order_count,ROUND(SUM(o.amount),2) revenue FROM orders o JOIN customers c ON o.customer_id=c.customer_id GROUP BY c.region ORDER BY revenue DESC",db))
```
