# Mushroom Classification

**Portfolio category:** Technical demonstration

Classification with preprocessing, Logistic Regression, Random Forest, ROC-AUC and feature importance using synthetic categorical data.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,RocCurveDisplay
rng=np.random.default_rng(10); n=1800
cap=rng.choice(['bell','conical','convex','flat'],n); color=rng.choice(['brown','gray','red','yellow','white'],n); odor=rng.choice(['none','almond','anise','foul'],n,p=[.48,.18,.14,.20]); habitat=rng.choice(['woods','grass','urban','leaves'],n)
p=1/(1+np.exp(-(-1+1.9*(odor=='foul')-.8*(odor=='almond')+.45*(color=='red')+.25*(habitat=='leaves'))))
target=rng.binomial(1,p)
df=pd.DataFrame({'cap_shape':cap,'cap_color':color,'odor':odor,'habitat':habitat,'target':target})
X=df.drop(columns='target'); y=df.target
cat=list(X.columns); prep=ColumnTransformer([('cat',Pipeline([('imp',SimpleImputer(strategy='most_frequent')),('oh',OneHotEncoder(handle_unknown='ignore'))]),cat)])
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
models={'Logistic Regression':Pipeline([('prep',prep),('model',LogisticRegression(max_iter=1000))]),'Random Forest':Pipeline([('prep',prep),('model',RandomForestClassifier(n_estimators=300,max_depth=8,random_state=42))])}
rows=[]
for name,m in models.items():
 m.fit(Xtr,ytr); pred=m.predict(Xte); prob=m.predict_proba(Xte)[:,1]; rows.append([name,accuracy_score(yte,pred),precision_score(yte,pred),recall_score(yte,pred),f1_score(yte,pred),roc_auc_score(yte,prob)])
results=pd.DataFrame(rows,columns=['Model','Accuracy','Precision','Recall','F1','ROC_AUC']); display(results.round(3))
fig,ax=plt.subplots(figsize=(7,5))
for name,m in models.items(): RocCurveDisplay.from_estimator(m,Xte,yte,ax=ax,name=name)
ax.set_title('Mushroom Classification ROC Comparison'); plt.tight_layout()
```
