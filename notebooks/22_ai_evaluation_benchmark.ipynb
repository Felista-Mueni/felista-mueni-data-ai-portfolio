# AI Response Evaluation Benchmark

Transparent example rubric and inter-rater agreement on fictional AI response assessments.

**Evidence status:** Reproducible synthetic/illustrative portfolio demonstration; not client data or externally validated outcomes.


```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.metrics import cohen_kappa_score
cases=pd.DataFrame({'case_id':range(1,13),'relevance':[5,4,3,5,2,4,5,3,4,2,5,3],'factuality':[5,3,2,4,2,4,5,3,4,1,5,3],'instruction_following':[5,4,3,5,3,4,4,3,5,2,5,4],'clarity':[4,5,4,5,3,4,5,4,4,3,5,4],'reviewer_a':[1,1,0,1,0,1,1,0,1,0,1,0],'reviewer_b':[1,0,0,1,0,1,1,1,1,0,1,0]})
dimensions=['relevance','factuality','instruction_following','clarity']; cases['quality_score']=cases[dimensions].mean(axis=1)
display(cases); print('Cohen kappa:',round(cohen_kappa_score(cases.reviewer_a,cases.reviewer_b),3))
cases[dimensions].mean().plot(kind='bar',title='Synthetic AI Quality Assessment',ylim=(0,5)); plt.tight_layout()
print('Ratings are fictional demonstrations, not real model benchmark results.')
```
