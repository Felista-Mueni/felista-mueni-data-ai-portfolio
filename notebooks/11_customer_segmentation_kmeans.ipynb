# Customer Segmentation with K-Means

**Portfolio category:** Professional technical demonstration

Unsupervised learning demonstration using synthetic RFM customer data, silhouette-score model selection, cluster profiling and business interpretation.

> This notebook uses simulated data and is clearly labelled as portfolio evidence.


```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
rng=np.random.default_rng(21); n=900; s=rng.choice(4,n,p=[.25,.30,.25,.20])
base=np.array([[15,28,420],[55,10,160],[95,4,65],[30,18,240]])
df=pd.DataFrame({'Recency':[max(1,base[x,0]+rng.normal(0,7)) for x in s],'Frequency':[max(1,base[x,1]+rng.normal(0,5)) for x in s],'Monetary':[max(20,base[x,2]+rng.normal(0,90)) for x in s]})
Z=StandardScaler().fit_transform(df); scores=[]
for k in range(2,9):
 km=KMeans(n_clusters=k,n_init=20,random_state=42); lab=km.fit_predict(Z); scores.append((k,silhouette_score(Z,lab)))
sc=pd.DataFrame(scores,columns=['k','silhouette']); display(sc.round(3))
best=int(sc.loc[sc.silhouette.idxmax(),'k']); km=KMeans(n_clusters=best,n_init=30,random_state=42); df['Cluster']=km.fit_predict(Z)
profile=df.groupby('Cluster')[['Recency','Frequency','Monetary']].mean().round(1); profile['Customers']=df.Cluster.value_counts().sort_index(); display(profile)
fig,ax=plt.subplots(figsize=(8,5))
for c,g in df.groupby('Cluster'): ax.scatter(g.Frequency,g.Monetary,s=18,alpha=.5,label=f'Cluster {c}')
ax.set_xlabel('Frequency'); ax.set_ylabel('Monetary value'); ax.set_title('Customer Segments'); ax.legend(); plt.tight_layout()
```
