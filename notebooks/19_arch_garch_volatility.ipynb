# ARCH/GARCH Volatility Modelling

**Portfolio category:** Technical demonstration

Estimate volatility persistence in simulated financial returns using an ARCH/GARCH model.

> This notebook uses simulated/synthetic data where stated and is portfolio evidence, not a claim of client results.


```python
import numpy as np,pandas as pd,matplotlib.pyplot as plt
from statsmodels.tsa.arch_model import arch_model
rng=np.random.default_rng(19); n=1500; eps=np.zeros(n); h=np.zeros(n); h[0]=.0002
for t in range(1,n):
 h[t]=.00002+.12*eps[t-1]**2+.82*h[t-1]; eps[t]=np.sqrt(h[t])*rng.standard_normal()
returns=pd.Series(eps,index=pd.date_range('2020-01-01',periods=n))
fig,ax=plt.subplots(figsize=(10,4)); returns.plot(ax=ax); ax.set_title('Simulated Returns'); plt.tight_layout()
model=arch_model(returns*100,vol='Garch',p=1,q=1,mean='Constant',dist='normal'); fit=model.fit(disp='off'); print(fit.summary())
vol=fit.conditional_volatility
fig,ax=plt.subplots(figsize=(10,4)); vol.plot(ax=ax); ax.set_title('Estimated Conditional Volatility'); plt.tight_layout()
print('Estimated parameters:'); display(fit.params.to_frame('estimate').round(4))
```
