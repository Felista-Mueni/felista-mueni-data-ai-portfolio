# NLP Text Classification

Reproducible text classification using synthetic example messages, TF-IDF, logistic regression and cross-validation.

**Evidence status:** Reproducible synthetic/illustrative portfolio demonstration; not client data or externally validated outcomes.


```python
import pandas as pd,numpy as np,matplotlib.pyplot as plt
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold,cross_val_score,train_test_split
from sklearn.metrics import classification_report,ConfusionMatrixDisplay
positive=['excellent service and fast delivery','very helpful support and great experience','reliable product and friendly team','quick response and smooth process','satisfied with the clear explanation']
negative=['poor service and delayed delivery','unhelpful support and bad experience','unreliable product and rude team','slow response and confusing process','disappointed with the unclear explanation']
rng=np.random.default_rng(23); rows=[]
for i in range(300):
 label=i%2; base=rng.choice(positive if label else negative); rows.append((base+' '+rng.choice(['today','overall','for my order','from this team']),label))
df=pd.DataFrame(rows,columns=['text','label']); Xtr,Xte,ytr,yte=train_test_split(df.text,df.label,stratify=df.label,test_size=.25,random_state=42)
model=Pipeline([('tfidf',TfidfVectorizer(ngram_range=(1,2))),('lr',LogisticRegression(max_iter=1000))])
cv=cross_val_score(model,Xtr,ytr,cv=StratifiedKFold(5,shuffle=True,random_state=42),scoring='f1'); model.fit(Xtr,ytr)
print('CV F1:',cv.mean().round(3),'SD:',cv.std().round(3)); print(classification_report(yte,model.predict(Xte)))
ConfusionMatrixDisplay.from_estimator(model,Xte,yte); plt.tight_layout()
print('This toy dataset illustrates a pipeline; it is not a generalizable real-world benchmark.')
```
